```python
# import os
# os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras.layers import Dense

c = np.array([-40, -10, 0, 8, 15, 22, 38])
f = np.array([-40, 14, 32, 46, 59, 72, 100])

model = keras.Sequential()
model.add(Dense(units=1, input_shape=(1,), activation='linear'))
model.compile(loss='mean_squared_error', optimizer=keras.optimizers.Adam(0.1))

history = model.fit(c, f, epochs=500, verbose=0)
print("Обучение завершено")

print(model.predict([100]))
print(model.get_weights())

plt.plot(history.history['loss'])
plt.grid(True)
plt.show()
```

```python
import os
# os — Python-un əməliyyat sistemi ilə işləməsi üçün standart kitabxanadır.
# Burada TensorFlow-un terminalda çıxardığı bəzi texniki mesajları idarə etmək üçün istifadə olunur.


os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'
# TF_CPP_MIN_LOG_LEVEL TensorFlow-un hansı mesajları göstərməsini müəyyən edir.
#
# '0' → bütün mesajları göstərir
# '1' → INFO mesajlarını gizlədir
# '2' → INFO və WARNING mesajlarını gizlədir
# '3' → INFO, WARNING və ERROR mesajlarını gizlədir
#
# '2' seçdiyimiz üçün yalnız daha ciddi xətalar görünəcək.
# Bu sətr TensorFlow import edilməzdən ƏVVƏL yazılmalıdır.


import numpy as np
# NumPy — ədədlər və massivlərlə işləmək üçün istifadə olunan kitabxanadır.
# Biz aşağıda Celsius və Fahrenheit qiymətlərini NumPy array kimi yaradacağıq.


import matplotlib.pyplot as plt
# matplotlib.pyplot — qrafik çəkmək üçün istifadə olunur.
# Aşağıda neyron şəbəkəsinin təlim zamanı "loss" dəyərinin necə azaldığını qrafikdə göstərəcəyik.


from tensorflow import keras
# TensorFlow daxilindəki Keras API-ni import edirik.
# Keras neyron şəbəkələri yaratmaq, öyrətmək və qiymətləndirmək üçün daha rahat interfeysdir.


from tensorflow.keras.layers import Dense
# Dense — tam əlaqəli (fully connected) neyron şəbəkə qatıdır.
# Bizim modelimizdə yalnız 1 ədəd Dense qatı olacaq.


c = np.array([-40, -10, 0, 8, 15, 22, 38])
# c — Celsius temperatur qiymətləridir.
#
# Bunlar modelə verilən GİRİŞ (input) məlumatlarıdır.
#
# Məsələn:
# -40°C
# -10°C
# 0°C
# 8°C
# 15°C
# 22°C
# 38°C


f = np.array([-40, 14, 32, 46, 59, 72, 100])
# f — həmin Celsius temperaturlarının Fahrenheit qarşılıqlarıdır.
#
# Bunlar modelin öyrənməli olduğu ÇIXIŞ (target/output) qiymətləridir.
#
# Məsələn:
# -40°C → -40°F
# -10°C → 14°F
# 0°C   → 32°F
# 8°C   → 46°F
# 15°C  → 59°F
# 22°C  → 72°F
# 38°C  → 100°F
#
# Yəni model c qiymətini görüb uyğun f qiymətini tapmağı öyrənəcək.


model = keras.Sequential()
# Sequential — qatları ardıcıl şəkildə bir-birinin arxasınca yerləşdirən model tipidir.
#
# Bizim modelimiz çox sadədir:
#
# Input → Dense → Output
#
# Yəni giriş məlumatı Dense qatına daxil olur və nəticə çıxış kimi alınır.


model.add(Dense(units=1, input_shape=(1,), activation='linear'))
# Dense — tam əlaqəli (fully connected) neyron şəbəkə qatıdır.
# Modelə bir Dense qatı əlavə edirik.
#
# units=1
# → Qatda yalnız 1 neyron olacaq.
# Çünki nəticədə biz yalnız 1 qiymət istəyirik:
# Celsius → 1 Fahrenheit qiyməti.
#
# input_shape=(1,)
# → Modelə hər dəfə 1 ədəd giriş veriləcəyini bildirir.
# Məsələn:
# 22 → model → 72
#
# activation='linear'
# → Aktivasiya funksiyası linear olacaq.
# Linear funksiya nəticəni başqa formada dəyişdirmir.
#
# Bu problem üçün linear aktivasiya məntiqlidir,
# çünki Celsius və Fahrenheit arasında xətti əlaqə var:
#
# F = C × 9/5 + 32
#
# Neyron şəbəkəsi bu əlaqəni öz çəkilərini (weight)
# və bias-ını öyrənərək tapmağa çalışacaq.


model.compile(
    loss='mean_squared_error',
    optimizer=keras.optimizers.Adam(0.1)
)
# compile() modelin necə öyrənəcəyini müəyyən edir.
#
# Burada iki əsas parametr veririk:
#
# 1. loss
# 2. optimizer
#
# loss='mean_squared_error'
# → Mean Squared Error (MSE) itki funksiyasıdır.
#
# Modelin proqnozu ilə real cavab arasındakı fərqi ölçür.
#
# Formula:
#
# MSE = (real - proqnoz)² orta qiyməti
#
# Məsələn:
# Real = 72
# Model = 70
#
# Xəta = 72 - 70 = 2
# Kvadrat xəta = 2² = 4
#
# Modelin məqsədi LOSS dəyərini mümkün qədər azaltmaqdır.
#
#
# optimizer=keras.optimizers.Adam(0.1)
# → Adam modelin çəkilərini dəyişdirərək səhvi azaltmağa çalışır.
#
# 0.1 → learning rate-dir (öyrənmə sürəti).
#
# Learning rate modelin hər addımda çəkiləri nə qədər dəyişəcəyini
# müəyyən edir.
#
# Böyük learning rate → daha sürətli, amma qeyri-sabit ola bilər.
# Kiçik learning rate → daha yavaş, amma daha stabil ola bilər.


history = model.fit(c, f, epochs=500, verbose=0)
# fit() — modelin öyrədilməsi prosesini başladır.
#
# c → giriş məlumatlarıdır (Celsius).
# f → düzgün cavablardır (Fahrenheit).
#
# epochs=500
# → Model bütün məlumatları 500 dəfə görəcək və öyrənəcək.
#
# 1 epoch = bütün training dataset-in modeldən bir dəfə keçirilməsi.
#
# verbose=0
# → Təlim zamanı hər epoch üçün ekrana məlumat çıxarmasın.
#
# history dəyişənində isə təlim prosesinin nəticələri saxlanılır.
# Məsələn, hər epoch-dakı loss qiymətlərini:
#
# history.history['loss']
#
# vasitəsilə əldə edə bilərik.


print("Обучение завершено")
# Ekrana "Təlim tamamlandı" yazdırır.
# Bu sətrin modelin öyrədilməsi ilə birbaşa əlaqəsi yoxdur.
# Sadəcə istifadəçiyə fit() prosesinin bitdiyini bildirir.


print(model.predict([100]))
# predict() — öyrədilmiş modeldən proqnoz almaq üçün istifadə olunur.
#
# Burada modelə 100 veririk.
#
# 100 → Celsius temperaturudur.
#
# Model isə onun Fahrenheit qarşılığını proqnozlaşdırmalıdır.
#
# Düzgün riyazi cavab:
#
# F = 100 × 9/5 + 32
# F = 180 + 32
# F = 212
#
# Buna görə model təxminən:
#
# [[212]]
#
# kimi nəticə verməlidir.
#
# Əgər nəticə 212-yə çox yaxındırsa,
# model Celsius → Fahrenheit əlaqəsini yaxşı öyrənib.


print(model.get_weights())
# get_weights() modelin öyrəndiyi WEIGHT və BIAS dəyərlərini göstərir.
#
# Bizim modelimizdə 1 neyron olduğu üçün:
#
# Weight (çəki)
# Bias (sürüşmə)
#
# olacaq.
#
# Modelin öyrəndiyi düstur təxminən belədir:
#
# output = weight × input + bias
#
# Fahrenheit üçün düzgün düstur:
#
# F = 1.8 × C + 32
#
# Buna görə modelin weight-i təxminən:
#
# 1.8
#
# bias-ı isə təxminən:
#
# 32
#
# olacaq.
#
# Yəni neyron şəbəkəsi əslində bu riyazi əlaqəni
# öz weight və bias parametrlərində öyrənmiş olur.


plt.plot(history.history['loss'])
# Təlim zamanı toplanmış loss qiymətlərini qrafikdə çəkir.
#
# history.history['loss']
# → hər epoch üçün loss qiymətlərinin siyahısıdır.
#
# Məsələn:
#
# Epoch 1   → böyük loss
# Epoch 2   → daha kiçik loss
# Epoch 3   → daha kiçik loss
# ...
# Epoch 500 → çox kiçik loss
#
# Qrafik aşağı doğru gedirsə,
# model adətən öyrənir və səhv azalır.


plt.grid(True)
# Qrafikə tor (grid) əlavə edir.
# Bu, qrafikdə qiymətləri daha rahat oxumağa kömək edir.


plt.show()
# Hazırlanmış qrafiki ekranda göstərir.
```

### ADAM

```python
myAdam = keras.optimizers.Adam(learning_rate=0.1)
# keras.optimizers.Adam(...) Adam optimizer sinifindən
# ayrıca bir obyekt yaradır.
#
# Adətən model.compile() daxilində sadəcə:
#
# optimizer='adam'
#
# yazırıq və Keras avtomatik olaraq default parametrlərlə
# (learning_rate=0.001) Adam optimizer yaradır.
#
# Lakin bəzən öyrənmə sürətini özümüz təyin etmək istəyirik.
# Bunun üçün Adam optimizerini ayrıca dəyişəndə yaradırıq.
#
# learning_rate=0.1
# → optimizerin öyrənmə sürətini bildirir.
#
# Learning rate modelin hər addımda weight-ləri
# nə qədər dəyişdirəcəyini müəyyən edir.
#
# Böyük learning rate (məsələn 0.1):
# → model daha böyük addımlarla öyrənir
# → öyrənmə sürətlənə bilər
# → amma model optimum nöqtəni "ötüb keçə" bilər,
#    loss stabil azalmaya bilər
#
# Kiçik learning rate (məsələn 0.0001):
# → model daha kiçik addımlarla öyrənir
# → öyrənmə yavaşdır
# → amma nəticə adətən daha stabildir
#
# Default Adam learning_rate dəyəri:
#
# 0.001
#
# Burada 0.1 dəyəri default-dan xeyli böyükdür,
# ona görə öyrənmə prosesi daha "aqressiv" olacaq.
#
# myAdam dəyişəni bu tənzimlənmiş optimizer obyektini saxlayır.
# Bu obyekti sonra model.compile() daxilində istifadə edəcəyik.


model.compile(
    optimizer=myAdam,
    # Burada 'adam' string-i yazmaq əvəzinə,
    # yuxarıda yaratdığımız myAdam obyektini veririk.
    #
    # Bu, modelin öyrənmə zamanı bizim təyin etdiyimiz
    # learning_rate=0.1 dəyərini istifadə edəcəyi mənasına gəlir.
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

### SGD

```python
mySGD = keras.optimizers.SGD(learning_rate=0.1, momentum=0.0, nesterov=True)
# keras.optimizers.SGD(...) — Stochastic Gradient Descent (SGD)
# optimizer sinifindən ayrıca bir obyekt yaradır.
#
# SGD — ən klassik və ən sadə optimizasiya alqoritmlərindən biridir.
#
# Adam-dan fərqli olaraq SGD adətən daha "sadə" məntiqlə işləyir:
# hər addımda gradient hesablanır və weight-lər
# birbaşa bu gradient istiqamətində yenilənir.
#
# Ümumi formulu:
#
# weight = weight - learning_rate * gradient
#
#
# learning_rate=0.1
# → optimizerin öyrənmə sürətini bildirir.
#
# Yəni model hər addımda weight-ləri
# nə qədər dəyişdirəcək, bunu göstərir.
#
# SGD üçün 0.1 nisbətən yüksək sayıla bilər,
# çünki SGD-nin default learning rate-i adətən
# daha kiçik (məsələn 0.01) olur.
#
#
# momentum=0.0
# → momentum SGD-yə "inertiya" effekti əlavə edən parametrdir.
#
# Momentum əvvəlki addımlardakı hərəkət istiqamətini
# nəzərə alaraq, öyrənməni daha hamar və sürətli edə bilər.
#
# Analoji olaraq:
# → təpədən aşağı diyirlənən topun sürəti
#    əvvəlki hərəkətindən təsirlənir.
#
# Burada momentum=0.0 olduğuna görə,
# bu əlavə "inertiya" effekti işə düşmür.
# Yəni sadə, "təmiz" SGD kimi davranır.
#
#
# nesterov=True
# → Nesterov Accelerated Gradient (NAG) adlı təkmilləşdirilmiş
# momentum texnikasını aktiv edir.
#
# Nesterov metodunda gradient hesablanmadan əvvəl,
# "irəlidəki" mövqe təxmini olaraq nəzərə alınır.
#
# Bu, adi momentumdan bir az daha "ağıllı" hesab olunur,
# çünki model növbəti addımı qabaqcadan görməyə çalışır.
#
# Lakin qeyd:
# → momentum=0.0 olduğu üçün, nesterov=True olsa belə,
#    praktikada real təsiri çox məhdud olacaq,
#    çünki nesterov mexanizmi momentum üzərində işləyir.
#
#
# mySGD dəyişəni bu tənzimlənmiş SGD optimizer obyektini saxlayır.
# Bu obyekti sonra model.compile() daxilində istifadə edəcəyik.


model.compile(
    optimizer=mySGD,
    # Burada 'sgd' string-i yazmaq əvəzinə,
    # yuxarıda yaratdığımız mySGD obyektini veririk.
    #
    # Beləliklə model öyrədilərkən Adam əvəzinə
    # bizim təyin etdiyimiz parametrlərlə (learning_rate=0.1,
    # momentum=0.0, nesterov=True) SGD optimizer istifadə olunacaq.
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

### Validation Manual Selcetion

```python
size_val = 10000
# size_val dəyişəni validation üçün ayıracağımız
# şəkillərin sayını göstərir.
#
# Burada 10 000 şəkli validation (yoxlama) datasetinə
# ayırmaq qərarına gəlmişik.
#
# Bu, validation_split=0.2 istifadə etməyin əl ilə
# (manual) edilən versiyasıdır.
#
# MNIST training datasetində ümumilikdə 60 000 şəkil
# olduğu üçün:
#
# 10 000 → validation üçün
# 50 000 → training üçün
#
# ayrılacaq.


x_val_split = x_train[:size_val]
# x_train datasetindən ilk 10 000 şəkli seçirik.
#
# [:size_val]
# → başlanğıcdan (0-cı indeksdən) size_val-a
#    (10000-ə) qədər olan hissəni götürür.
#
# Yəni:
#
# x_train[0]    → daxildir
# x_train[1]    → daxildir
# ...
# x_train[9999] → daxildir
# x_train[10000] → daxil deyil
#
# Bu seçilmiş 10 000 şəkil validation üçün
# giriş məlumatı olacaq.
#
# x_val_split-in forması:
#
# (10000, 28, 28)


y_val_split = y_train_cat[:size_val]
# y_train_cat-dan da eyni indekslərə uyğun (ilk 10 000)
# cavabları seçirik.
#
# Bu çox vacibdir:
# → x_val_split və y_val_split eyni indekslərə
#    uyğun olmalıdır ki, hər şəklin öz düzgün
#    cavabı ilə uyğunlaşsın.
#
# y_train_cat artıq One-Hot Encoding formasında
# olduğu üçün y_val_split də One-Hot formada olacaq.
#
# y_val_split-in forması:
#
# (10000, 10)


x_train_split = x_train[size_val:]
# x_train datasetindən 10 000-ci indeksdən
# sona qədər olan hissəni seçirik.
#
# [size_val:]
# → size_val-dan (10000-dən) başlayaraq
#    sona qədər olan bütün elementləri götürür.
#
# Yəni:
#
# x_train[10000] → daxildir
# x_train[10001] → daxildir
# ...
# x_train[59999] → daxildir
#
# Bu, artıq validation üçün ayrılmayan,
# yalnız training üçün istifadə olunacaq şəkillərdir.
#
# x_train_split-in forması:
#
# (50000, 28, 28)
#
# Çünki:
#
# 60000 - 10000 = 50000


y_train_split = y_train_cat[size_val:]
# y_train_cat-dan da eyni şəkildə 10000-ci indeksdən
# sona qədər olan cavabları seçirik.
#
# Bu, x_train_split-ə uyğun düzgün cavablardır.
#
# y_train_split-in forması:
#
# (50000, 10)
#
#
# Nəticədə əldə etdiyimiz 4 dəyişən:
#
# x_train_split, y_train_split
# → modelin öyrənməsi üçün (50 000 şəkil)
#
# x_val_split, y_val_split
# → modelin təlim zamanı yoxlanılması üçün (10 000 şəkil)
#
# Bu bölgü sayəsində, əvvəlki nümunədəki kimi
# validation_split=0.2 yazmaq əvəzinə,
# validation datasını özümüz sabit və əl ilə
# ayırmış oluruq. Bunun üstünlüyü ondadır ki,
# hər dəfə eyni validation set istifadə olunur
# (təsadüfi random bölgü olmur).
```

```python
model.fit(x_train_split, y_train_split, batch_size=32, epochs=5, validation_data=(x_val_split, y_val_split))
# model.fit() modelin öyrədilməsini başladır.
#
# İndi artıq x_train_split, y_train_split, x_val_split
# və y_val_split dəyişənlərini əvvəlcədən əl ilə
# hazırlamışıq, ona görə validation_split=0.2 yazmağa
# ehtiyac yoxdur — validation datasını birbaşa özümüz veririk.
#
#
# x_train_split
# → modelin öyrənməsi üçün istifadə olunan giriş şəkilləridir
#    (50 000 şəkil, x_train-in 10000-ci indeksdən sonrakı hissəsi).
#
#
# y_train_split
# → x_train_split-ə uyğun düzgün cavablardır (One-Hot formada,
#    50 000 nümunə).
#
#
# batch_size=32
# → model məlumatları 32-lik qruplar (batch) şəklində emal edir.
#
# Yəni bütün 50 000 şəkil bir dəfəyə deyil,
# 32 nümunədən ibarət kiçik hissələr şəklində modeldən keçirilir
# və hər batch-dan sonra weight-lər yenilənir.
#
#
# epochs=5
# → bütün training datasetinin (x_train_split) modeldən
# 5 dəfə keçiriləcəyini bildirir.
#
# 1 epoch = bütün training datasetinin
# bir dəfə tam şəkildə model tərəfindən görülməsi.
#
#
# validation_data=(x_val_split, y_val_split)
# → əvvəlcədən əl ilə ayırdığımız 10 000 şəkillik
# validation datasetini birbaşa veririk.
#
# x_val_split → validation üçün giriş şəkilləri
# y_val_split → validation üçün düzgün cavablar
#
# Bunun validation_split=0.2-dən fərqi:
# → validation_split istifadə etsəydik, Keras hər dəfə
#    training datasetinin son 20%-ni avtomatik ayırardı.
#
# → Burada isə validation setini özümüz əvvəlcədən
#    sabit şəkildə müəyyən etmişik, ona görə hər çalışdırdıqda
#    eyni 10 000 şəkil validation üçün istifadə olunacaq.
#
#
# Təlim prosesi zamanı model:
#
# 1. x_train_split (50 000 şəkil) üzərində öyrənir
# 2. hər epoch sonunda x_val_split (10 000 şəkil) üzərində yoxlanılır
# 3. training və validation loss/accuracy nəticələri
#    ekranda göstərilir
#
# Bu, modelin yeni (görmədiyi) məlumatlar üzərində
# necə performans göstərdiyini izləməyə kömək edir və
# overfitting-i (modelin yalnız training datasını
# "əzbərləməsini") aşkar etməyə imkan verir.
```

###  Validation selection with 
- from sklearn.model_selection import train_test_split

```python
from sklearn.model_selection import train_test_split
# sklearn (scikit-learn) — machine learning üçün istifadə olunan
# məşhur Python kitabxanasıdır.
#
# model_selection modulu daxilində dataset bölmək,
# model seçmək, cross-validation etmək üçün funksiyalar var.
#
# train_test_split — bunlardan biridir və bizə datasetimizi
# training və validation (və ya test) hissələrinə
# TƏSADÜFİ (random) şəkildə bölməyə imkan verir.
#
# Bu importu TensorFlow/Keras importlarından ayrı,
# faylın əvvəlində etmək daha doğrudur.


x_train_split, x_val_split, y_train_split, y_val_split = train_test_split(x_train, y_train_cat, test_size=0.2)
# train_test_split() funksiyasını çağırırıq.
#
# Bu funksiya bir datasetin daxilindəki nümunələri
# TƏSADÜFİ şəkildə iki hissəyə bölür:
#
# 1. Training (öyrənmə) hissəsi
# 2. Validation (yoxlama) hissəsi
#
# Bu, əvvəlki üsuldan (x_train[:size_val] kimi əl ilə,
# sıra ilə bölmə) fərqlənir — burada bölgü TƏSADÜFİ olur.
#
#
# x_train
# → funksiyaya verdiyimiz giriş məlumatı (60 000 şəkil).
#
#
# y_train_cat
# → x_train-ə uyğun One-Hot Encoding formasında cavablar.
#
#
# test_size=0.2
# → datasetin 20%-nin validation üçün ayrılacağını bildirir.
#
# Yəni:
#
# 80% → training üçün (təxminən 48 000 şəkil)
# 20% → validation üçün (təxminən 12 000 şəkil)
#
#
# Funksiya 4 dəyər qaytarır və biz onları eyni sırada
# 4 dəyişənə mənimsədirik:
#
# x_train_split → training üçün giriş şəkilləri
# x_val_split   → validation üçün giriş şəkilləri
# y_train_split → training üçün düzgün cavablar
# y_val_split   → validation üçün düzgün cavablar
#
# DİQQƏT: sıra vacibdir!
# → sklearn həmişə: (train_giriş, test_giriş, train_cavab, test_cavab)
#    formasında qaytarır.
#
#
# Bu üsulun əvvəlki (əl ilə bölmə) üsuldan üstünlüyü:
# → bölgü təsadüfi olduğu üçün, dataset əvvəlcədən
#    hər hansı bir sıra ilə düzülmüş olsa belə
#    (məsələn əvvəlcə hamısı "0" rəqəmləri, sonra "1"lər və s.),
#    training və validation setlərinin hər ikisi
#    bütün siniflərdən qarışıq nümunələr alacaq.


model.fit(x_train_split, y_train_split, batch_size=32, epochs=5, validation_data=(x_val_split, y_val_split))
# model.fit() modelin öyrədilməsini başladır.
#
# x_train_split
# → train_test_split() ilə ayrılmış training giriş şəkilləri.
#
# y_train_split
# → onlara uyğun düzgün cavablar (One-Hot formada).
#
# batch_size=32
# → model məlumatları 32-lik qruplar (batch) şəklində emal edir,
#    hər batch-dan sonra weight-lər yenilənir.
#
# epochs=5
# → bütün training datasetinin modeldən 5 dəfə keçiriləcəyini bildirir.
#
# validation_data=(x_val_split, y_val_split)
# → train_test_split() ilə ayrılmış validation datasını
#    birbaşa veririk.
#
# Model hər epoch-dan sonra:
# 1. training datası üzərində öyrənməyə davam edir
# 2. validation datası üzərində yoxlanılır (bu datadan
#    weight-lər dəyişdirilmir, sadəcə nəzarət üçündür)
# 3. training və validation loss/accuracy nəticələri
#    ekranda göstərilir
#
# Bu, modelin görmədiyi məlumat üzərində real performansını
# izləməyə və overfitting-i vaxtında aşkar etməyə kömək edir.
```

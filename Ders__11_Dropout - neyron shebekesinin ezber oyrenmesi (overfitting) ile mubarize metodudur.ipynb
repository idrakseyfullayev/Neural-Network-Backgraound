```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist         # библиотека базы выборок Mnist
from tensorflow import keras
from tensorflow.keras.layers import Dense, Flatten, Dropout

(x_train, y_train), (x_test, y_test) = mnist.load_data()

# стандартизация входных данных
x_train = x_train / 255
x_test = x_test / 255

y_train_cat = keras.utils.to_categorical(y_train, 10)
y_test_cat = keras.utils.to_categorical(y_test, 10)

limit = 5000
x_train_data = x_train[:limit]
y_train_data = y_train_cat[:limit]

x_valid = x_train[limit:limit*2]
y_valid = y_train_cat[limit:limit*2]

model = keras.Sequential([
    Flatten(input_shape=(28, 28, 1)),
    Dense(300, activation='relu'),
    Dropout(0.8),
    Dense(10, activation='softmax') ])

# print(model.summary())      # вывод структуры НС в консоль

model.compile(optimizer='adam',
             loss='categorical_crossentropy',
             metrics=['accuracy'])


his = model.fit(x_train_data, y_train_data, epochs=50, batch_size=32, validation_data=(x_valid, y_valid))


plt.plot(his.history['loss'])
plt.plot(his.history['val_loss'])
plt.show()
```

```python
model = keras.Sequential([
    Flatten(input_shape=(28, 28, 1)),
    Dense(300, activation='relu'),
    Dropout(0.8),
    # Dropout — overfitting-in (modelin training datasını
    # "əzbərləməsinin") qarşısını almaq üçün istifadə olunan
    # bir regularization (tənzimləmə) texnikasıdır.
    #
    # Dropout(0.8) o deməkdir ki, hər təlim addımında
    # (hər forward/backward pass zamanı) əvvəlki Dense(300)
    # qatındakı neyronların TƏSADÜFİ olaraq 80%-i
    # müvəqqəti söndürülür (deaktiv edilir).
    #
    # 0.8 → "dropout rate", yəni söndürülmə nisbətini bildirir.
    #
    # Beləliklə hər addımda:
    #
    # 300 neyrondan → təxminən 240-ı söndürülür
    #                  (300 × 0.8 = 240)
    # → təxminən 60-ı isə fəal qalır
    #                  (300 × 0.2 = 60)
    #
    # Hansı neyronların söndürüləcəyi hər dəfə TƏSADÜFİ
    # seçilir, ona görə model heç vaxt eyni neyron
    # kombinasiyasına arxalana bilmir.
    #
    # Niyə bu faydalıdır?
    # → Əgər dropout olmasaydı, bəzi neyronlar training
    #    datasını "əzbərləyərək" çox dominant ola bilərdi.
    #
    # → Dropout modeli hər dəfə fərqli (kiçildilmiş) bir
    #    şəbəkə kimi işləməyə məcbur edir. Bu, modelin
    #    daha ümumiləşdirilmiş (generalized) xüsusiyyətlər
    #    öyrənməsinə kömək edir, çünki heç bir tək neyrona
    #    "güvənə" bilmir.
    #
    # Analoji: komanda daxilində hər məşqdə 80% oyunçunun
    # təsadüfi olaraq çıxarılması kimi düşünə bilərsiniz —
    # qalan oyunçular komandanın işini tək başına yerinə
    # yetirməyi öyrənməli olur, nəticədə komanda bir neçə
    # "ulduz" oyunçudan asılı olmur.
    #
    # QEYD: 0.8 kifayət qədər YÜKSƏK bir dropout dəyəridir
    # (adətən 0.2–0.5 arası daha çox istifadə olunur).
    # Bu qədər yüksək dropout modelin öyrənməsini
    # xeyli çətinləşdirə bilər, çünki hər addımda
    # neyronların yalnız 20%-i fəal qalır.
    #
    # QEYD 2: Dropout YALNIZ təlim (training) zamanı işləyir.
    # Test/proqnoz (model.predict() və ya model.evaluate())
    # zamanı Dropout avtomatik olaraq deaktiv olur və
    # bütün neyronlar iştirak edir.

    Dense(10, activation='softmax')
])
```
